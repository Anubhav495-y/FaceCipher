```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Aim is to predict the marks of students of the test data
```

```python
# Use the file namd 'training data' to train the model

data = pd.read_excel('Training data.xlsx')
x_train = np.array(data.iloc[:,0:8])
y_train = np.array(data.iloc[:,8]).reshape(-1,1)

# Try plotting y_train with different features
# To get an idea whether to add some features or not
# Add some features if required in x_train
for i in range(8):
    plt.figure()
    plt.scatter(x_train[:, i], y_train)
    plt.xlabel(f'{data.columns[i]}')
    plt.ylabel('Marks')
    plt.title(f'Marks vs {data.columns[i]}')
    plt.show()                               # There are many patterns visible from the scatter plots

# Also do label encoding for features not represented in numbers
# refer the link if not know : https://youtu.be/589nCGeWG1w?si=t2Wa7LgbUOO4RooM

def feature_changing(x_train):
  # ---------
    # Your code here
    # internet: yes/no → 1/0
    x_train[:, 0] = np.where(x_train[:, 0] == 'yes', 1, 0)

    # sex: M/F → 0/1
    x_train[:, 1] = np.where(x_train[:, 1] == 'F', 1, 0)

    # convert everything to float
    x_train = x_train.astype(float)

    # Feature engineering (non-linear effects)
    studytime_sq = (x_train[:, 3] ** 2).reshape(-1, 1)
    absences_sq = (x_train[:, 5] ** 2).reshape(-1, 1)

    # concatenate original + new features
    x_train = np.concatenate((x_train, studytime_sq, absences_sq), axis=1)
  # ---------
    return x_train

x_train = feature_changing(x_train)
```

```python
def z_score(x_train):

  # ---------
    # write the code for feature scaling here
    # Your code here
     # mean of each feature (column)
    x_mean = np.mean(x_train, axis=0)

    # standard deviation of each feature (column)
    x_std = np.std(x_train, axis=0)

    # z-score normalization
    x_train = (x_train - x_mean) / x_std
  # ---------

    return x_train,x_std,x_mean
```

```python
def cost(x_train,y_train,w,b):

  # ---------
    # Your code here
    # Use mean square error as cost function
    # return cost
    m = x_train.shape[0]

    # y_hat
    y_pred = np.dot(x_train, w) + b

    # mean squared error loss
    loss = (1 / (2 * m)) * np.sum((y_pred - y_train) ** 2)
  # ---------

    return loss
```

```python
def gradient_descent(x_train,y_train,w,b):

  # ---------
    # Your code here
    # Choose learning rate yourself
    m = x_train.shape[0]
    alpha = 0.01          # learning rate
    iterations = 1000     # number of steps

    for _ in range(iterations):

        # predictions
        y_pred = np.dot(x_train, w) + b

        # gradients
        dw = (1 / m) * np.dot(x_train.T, (y_pred - y_train))
        db = (1 / m) * np.sum(y_pred - y_train)

        # update parameters
        w = w - alpha * dw
        b = b - alpha * db
  # ---------

    return w,b
```

```python
x_train = x_train.astype(np.float64)
x_train,x_std,x_mean = z_score(x_train)

np.random.seed(2147483647)
w = np.random.randn(x_train.shape[1],1)
b = np.random.randn(1)

old_cost = 0

while abs(old_cost - cost(x_train,y_train,w,b))>0.00001:
  old_cost = cost(x_train,y_train,w,b)
  w,b = gradient_descent(x_train,y_train,w,b)

x_predict = pd.read_excel('Test data.xlsx').iloc[:,:8].to_numpy()
x_predict = feature_changing(x_predict)
x_predict = (x_predict - x_mean)/x_std
ans = pd.read_excel('Test data.xlsx').iloc[:,8].to_numpy()

y_predict = np.dot(x_predict,w) + b

accuracy = 0
for dim in range(len(ans)):
  if abs(y_predict[dim]-ans[dim])<0.5: # do not change the tolerance as you'll be checked on +- 0.5 error only
    accuracy += 1
accuracy = round(accuracy*100/200.0,2)
ok = 'Congratulations' if accuracy>95 else 'Optimization required'
print(f"{ok}, your accuracy is {accuracy}%")
```
