## Pandas Implementation

```python
import pandas as pd

## Create a DataFrame from a dictionary
data = {
    'Name': ['Ramesh', 'Mahesh', 'Suresh'],
    'Age': [25, 30, 35],
    'City': ['Bangalore', 'Mumbai', 'Delhi']
}
df = pd.DataFrame(data)

#Display the first 2 rows of the data frame
print(df.head(2))

#Print the age column
print(df['Age'])  #df.Age() ??

#Filter rows where age is greater than 26
df1 = df.loc[df['Age']>26]
print(df1)

#Add a new column 'Country' with the value 'India' for all rows
df['Country'] = 'India'

data1 = {
    'Name': ['Ramesh', 'Mahesh', 'Suresh'],
    'Age': [25, None, 35],
    'City': ['Bangalore', 'Mumbai', 'Delhi']
}

df2= pd.DataFrame(data1)

# Fill missing values in the 'Age' column with the mean age
df2.fillna({'Age':df2['Age'].mean()},inplace = True)
df2
```

## Matplotlib Implementation

```python
import numpy as np
import matplotlib.pyplot as plt

xpoints = np.array([1, 2, 3, 4])
ypoints = np.array([2, 4, 6, 8])

# Plot these points without drawing a line
plt.scatter(xpoints, ypoints)

# Plotting with marker: Star marker
plt.plot(xpoints, ypoints, marker='*', linestyle='')

# Using fmt format: circular marker, red color, dashed line
plt.plot(xpoints, ypoints, 'or--')   # o = circle, r = red, -- = dashed

# Add xlabel, ylabel, title
plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Plotting Example")

# Create a scatter plot for xpoints and ypoints
# Set color to the scatter plot (Blue, Green, Red, Yellow)
colors = ['blue', 'green', 'red', 'yellow']
plt.scatter(xpoints, ypoints, c=colors, s=100)  # s= marker size

plt.show()
```
