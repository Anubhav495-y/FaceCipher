```python
#feature mapping
def map_feature(X1,X2):
  X1=np.atleast_1d(X1)  # to avoide shape error later, keeping both 1D
  X2=np.atleast_1d(X2)
  degree=6              # polynomial features upto 6th degree
  out=[]                # list: to store new features
  for i in range(1,degree+1):
    for j in range(i+1):
      out.append((X1**(i-j) * (X2**j))) # all polynomial outputs (X1)^a(X2)^b
  return np.stack(out, axis=1) # Vstack, makes a 2D array
```

```python
# @title Default title text

def compute_cost(X,y,w,b,lambda_=1):
#write your code
    m = X.shape[0]
    z = X @ w + b
    y_hat = 1 / (1 + np.exp(-z))

    loss = y*np.log(y_hat) + (1 - y)*np.log(1 - y_hat)
    total_cost = -(1/m) * np.sum(loss)
       
    return total_cost
```

```python
# @title Default title text
#compute regularised cost
def compute_cost_reg(X, y, w, b, lambda_ =1):
#write your code
  m = X.shape[0]
  z = X@w + b
  y_hat = 1 / (1 + np.exp(-z))
  
  loss = y*np.log(y_hat) + (1 - y)*np.log(1 - y_hat)

  total_cost = -(1/m)*np.sum(loss) + (lambda_/(2*m))*np.sum(w**2)
  
  return total_cost
```

```python
#compute gradient
def compute_gradient(X, y, w, b, lambda_=None):
#write your code
  m = X.shape[0]
  z = X@w + b
  y_hat = 1 / (1 + np.exp(-z))

  dj_db = (1/m) * np.sum(y_hat - y)
  dj_dw = (1/m) * (X.T @ (y_hat - y)) # to make (n,m) x (m,)

  return dj_db,dj_dw
```

```python
#plot decision boundary
def plot_decision_boundary(w,b,X,y):

    plot_data(X[:, 0:2], y)

    if X.shape[1] <= 2:
        plot_x = np.array([min(X[:, 0]), max(X[:, 0])])
        plot_y = (-1. / w[1]) * (w[0] * plot_x + b)

        plt.plot(plot_x, plot_y, c="b")

    else:
        u = np.linspace(-1, 1.5, 50)
        v = np.linspace(-1, 1.5, 50)

        z = np.zeros((len(u), len(v)))


        for i in range(len(u)):
            for j in range(len(v)):
                z[i,j] = sigmoid(np.dot(map_feature(u[i], v[j]), w) + b)[0]


        z = z.T


        plt.contour(u,v,z, levels = [0.5], colors="g")
```

```python
#predict your training model
def predict(X, w, b):
  m,n=X.shape
  p=np.zeros(m)
  #write your code
  z= X@w + b
  threshold = 0.6
  f = sigmoid(z)
  for i in range(m):
    if  (f[i] >= threshold):
      p[i]=1
    else : p[i] = 0  
  



  return p
```
